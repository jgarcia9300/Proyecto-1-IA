```python
# -------------------------------------------------------------
# Apriori con parámetros configurables de soporte, confianza y lift
# -------------------------------------------------------------

import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python

# -------------------------------------------------------------
# 1. Datos de ejemplo (tus transacciones)
# -------------------------------------------------------------
transactions = [
    ["leche", "pan", "huevo"],
    ["cereal", "leche", "crema", "huevo"],
    ["leche", "crema", "pan", "huevo"]
]
```

```python
# -------------------------------------------------------------
# 2. Parámetros configurables
# -------------------------------------------------------------
min_support = 0.6       # soporte mínimo
min_confidence = 0.6    # confianza mínima
min_lift = 1.0          # lift mínimo

print("Usando parámetros:")
print(f"min_support   = {min_support}")
print(f"min_confidence = {min_confidence}")
print(f"min_lift       = {min_lift}")
```

```python
# -------------------------------------------------------------
# 3. Codificación de las transacciones
# -------------------------------------------------------------
te = TransactionEncoder()
te_result = te.fit(transactions).transform(transactions)
df = pd.DataFrame(te_result, columns=te.columns_)
```

```python
# -------------------------------------------------------------
# 4. Cálculo de los itemsets frecuentes
# -------------------------------------------------------------
frequent_itemsets = apriori(df, min_support=min_support, use_colnames=True)

print("\nItemsets frecuentes:")
display(frequent_itemsets)
```

```python
# -------------------------------------------------------------
# 5. Cálculo de reglas de asociación con los parámetros
# -------------------------------------------------------------
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=min_confidence
)
# Filtro adicional por lift
rules = rules[rules["lift"] >= min_lift]
```

```python
# -------------------------------------------------------------
# 6. Mostrar las reglas resultantes
# -------------------------------------------------------------
print("\nReglas de asociación filtradas:")
rules_to_show = rules[["antecedents", "consequents", "support", "confidence", "lift"]]
display(rules_to_show)
```
